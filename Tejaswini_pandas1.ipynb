### Q. Student Assessment Data Cleaning

**Question:** A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing.

**Using Pandas, perform the following cleaning operations in sequence:**

### 1. Load the CSV file and inspect the dataset using `head()`, `info()`, and `isnull().sum()`.

````python
import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv")

print(df.head())
print(df.info())
print(df.isnull().sum())
````

### Explanation

`pd.read_csv()` loads the CSV file into a Pandas DataFrame. `head()` displays the first rows, `info()` shows the structure and data types, and `isnull().sum()` shows the number of missing values in each column.

### 2. Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

````python
df["Department"] = df["Department"].replace({
    "it": "IT",
    "I.T.": "IT",
    "finance": "Finance"
})

print(df["Department"])
````

### Explanation

The department names are standardized so that `IT`, `it`, and `I.T.` are represented as `IT`. Similarly, `finance` is changed to `Finance`. This makes the department values consistent for analysis.

### 3. Identify and remove exact duplicate student assessment records.

````python
df = df.drop_duplicates()

print(df)
````

### Explanation

`drop_duplicates()` removes exact duplicate rows from the DataFrame. In this dataset, the duplicate record for student `S102` is removed.

### 4. Convert the Marks column into a numeric format, handling `"88 marks"`, `"Absent"`, `"NA"`, and blank entries appropriately.

````python
df["Marks"] = pd.to_numeric(
    df["Marks"].astype(str).str.replace(" marks", "", regex=False),
    errors="coerce"
)

print(df["Marks"])
````

### Explanation

The Marks column is converted into a numeric format. The text `88 marks` is converted to `88`, while `Absent`, `NA`, and blank values are treated as missing values.

### 5. Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

````python
print(df[df["Marks"].isnull()][["Student_ID", "Student_Name", "Marks"]])
````

### Explanation

The missing marks should be investigated because `Absent` may mean the student did not attend the assessment, while blank or `NA` may mean the marks were not entered.

**Risk of deleting missing values:**  
Deleting missing values without investigating may remove valid information and may incorrectly remove students who were absent.

### 6. Identify marks outside the valid range of 0–100 and decide how they should be handled.

````python
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print(invalid_marks)
````

### Explanation

Marks should be within the valid range of 0 to 100. The value `105` is outside this range, so it is identified as an invalid mark.

The invalid value should be investigated before deciding whether to correct or remove it.

### 7. Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

````python
print(df)

print("Records removed:", 1)
print("Missing marks remaining:", df["Marks"].isnull().sum())
print("Invalid marks identified:", len(invalid_marks))
````

### Explanation

The cleaned DataFrame is displayed after removing the duplicate record and standardizing the department names.

- **Records removed:** 1 duplicate record
- **Missing marks remaining:** Marks that are blank, `Absent`, or `NA` remain missing.
- **Invalid marks identified:** `105` is outside the valid range of 0–100 and should be investigated.

The final report summarizes the cleaning results.
